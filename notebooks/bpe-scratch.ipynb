```python
text = "The cat is lazy"
vocab_size = 270
num_merges = vocab_size - 256
```

```python
text_bytes = text.encode("utf-8")
```

```python
ids = list(text_bytes)
```

```python
ids
```

```python
def get_stats(ids, counts=None):
    # [1,2,3,1,2] -> {(1,2):2,(2,3):1,(3,1):1}
    counts = {} if counts is None else counts
    # print(ids, ids[1:])
    for pair in zip(ids, ids[1:]):
        counts[pair] = counts.get(pair, 0) + 1
    return counts
```

```python
def merge(ids, pair, idx):
    # ids: [1,2,3,1,2]
    # pair: (1,2)
    # ix: 4
    newids = []
    i=0
    while i<len(ids):
        if ids[i]==pair[0] and i+1 < len(ids) -1 and ids[i+1]==pair[1]:
            newids.append(idx)
            i+=2
        else:
            newids.append(ids[i])
            i+=1
    return newids
```

```python
merges = {}
vocab = {idx: bytes([idx]) for idx in range(256)}
print(vocab)
for i in range(num_merges):
    stats = get_stats(ids)
    # print(stats)
    pair = max(stats, key=stats.get)
    # print(pair)
    idx = 256 + i
    ids = merge(ids, pair, idx)
    merges[pair] = idx
    vocab[idx] = vocab[pair[0]] + vocab[pair[1]]
    
print(merges)
```
